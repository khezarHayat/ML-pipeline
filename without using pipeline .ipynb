```python
import numpy as np
import pandas as pd
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier
```

```python
df=pd.read_csv('Titanic.csv')
```

```python
df
```

```python
df.head()
```

```python
df.drop(columns=['PassengerId','Name','Ticket','Cabin'],inplace=True)
```

```python
df
```

```python
## train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(df.drop(columns=['Survived']),df['Survived'],test_size=0.3,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
df.isnull().sum()
```

```python
si=SimpleImputer()
si_em=SimpleImputer(strategy='most_frequent')                        # in this funtion have inside another parameter called strategey which change the value you want on place of null value 
x_train_age=si.fit_transform(x_train[['Age']])
x_test_age=si.fit_transform(x_test[['Age']])
x_train_em=si_em.fit_transform(x_train[['Embarked']])
x_test_em=si_em.fit_transform(x_test[['Embarked']])
```

```python
x_test_age.shape
```

```python
x_train_em.shape
```

```python
# 2nd tranformation we have catagorical col also so we used sex,embarked
ohe_sex=OneHotEncoder(handle_unknown='ignore',sparse_output=False)
ohe_emb=OneHotEncoder(handle_unknown='ignore',sparse_output=False)
x_train_sex=ohe_sex.fit_transform(x_train[['Sex']])
x_test_sex=ohe_sex.fit_transform(x_test[['Sex']])
x_train_emb=ohe_emb.fit_transform(x_train_em)
x_test_emb=ohe_emb.fit_transform(x_test_em)
```

```python
x_train_sex.shape
```

```python
x_train_emb.dtype
```

```python
x_train_em.dtype
```

```python
# now we combine the all numpy array
x_train_rem=x_train.drop(columns=['Sex','Age','Embarked'])
x_test_rem=x_test.drop(columns=['Sex','Age','Embarked'])
```

```python
x_train_comb=np.concatenate((x_train_rem,x_train_age,x_train_sex,x_train_emb),axis=1)
x_test_comb=np.concatenate((x_test_rem,x_test_age,x_test_sex,x_test_emb),axis=1)
```

```python
clf=DecisionTreeClassifier()
clf.fit(x_train_comb,y_train)
```

```python
y_pred=clf.predict(x_test_comb)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
